```python
import chromadb

client = chromadb.Client()
```

```python
collections = client.create_collection("news_article")
```

```python
collections.add(
    ids=["1", "2", "3", "4"],
    documents=[
        "Apple is leading in a smart phone game with iphone sales up by 20%.",
        "Tesla booked a minor profit in the last quarter, but is still struggling to meet production targets.",
        "Apples are a great source of fiber and vitamin C.",
        "SpaceX got NASA contract worth $10 billion." 
    ])
```

```python
# Fetch documents and embeddings from the collection
data = collections.get(
    include=["documents", "embeddings", "metadatas"]
)

# print embeddings
for i, embedding in enumerate(data['embeddings']):
    print(f"Document {i+1}")
    print(f"Text: {data['documents'][i]}")
    print(f"Embedding Length: {len(embedding)}")
    print(f"First 10 dimensions of the embedding: {embedding[:10]}")
```

```python
results = collections.query(
    query_texts=["This is about Elon Musk"],
    n_results=2
)
print(results)
```

```python
results = collections.query(
    query_texts=["I need to find information about Apple and its products."],
    n_results=2
)
print(results)
```
