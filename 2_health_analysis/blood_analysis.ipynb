```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
llm = ChatGoogleGenerativeAI(model='gemini-3.6-flash')
```

```python
with open("blood_work.txt", "r") as f:
          blood_report = f.read()

print(blood_report[:200])
```

```python
extraction_prompt = f"""
You are a medical data extraction assistant.

From the blood report below, extract ALL test values and classify each one as HIGH, LOW, or NORMAL based on the reference ranges provided in the report.

Format your response as:
- Test Name: value | Status: HIGH/LOW/NORMAL | Reference: range

Blood Report:
{blood_report}
"""
```

```python
extracted_response = llm.invoke(extraction_prompt)
extraction_values = extracted_response.text

print("--- STAGE 1: EXTRACTED VALUES ---")
print(extraction_values)
```

```python
diet_prompt = f"""
You are a clinical nutrionist specializing in Nigerian dietary habit.

From the blood report below, extract ALL test values and classify each one as HIGH, LOW, or NORMAL based on the reference ranges provided in the report.

Based on the blood work analysis below write:
1. A short health summary in 3-5 lines explaining the patient's condition in simple language
2. A short, practical Nigerian diet plan having only two sections (1) Foods to avoid (2) Foods to eat more of.
    Do not include any other sections in the diet plan.

Blood Work Analysis:
{extraction_values}
"""
```

```python
diet_response = llm.invoke(diet_prompt)
diet_values = diet_response.text

print("--- STAGE 2: HEALTH SUMMARY AND DIET PLAN ---")
print(diet_values)
```
